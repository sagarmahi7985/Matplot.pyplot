### SUBPLOTS AND LAYOUT ADJUSTMENTS

```python
import matplotlib.pyplot as plt
```

```python
#plt.subplot(nrows, ncolumns, index)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
plt.subplot(1,2,1)   #1 row, 2 column, 1st subplot
plt.plot(x,y)
plt.title('Line Chart')

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title('Bar Chart')


plt.show
```

```python
import matplotlib.pyplot as plt
#fig, ax = plt.subplots(nrows, ncols, figesize=(width, height))
fig, ax = plt.subplots(1, 2, figsize = (10,5))

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

ax[0].plot(x,y, color='green')
ax[0].set_title('Line Plot')

ax[1].bar(x,y, color='red')
ax[1].set_title('Bar Chart')

fig.suptitle('Comparison of Line and Bar Charts')

plt.tight_layout()
plt.show()
```

### Saving figures and pictures

```python
#savefig('filename_extension', dpi = value, bbox_inches = 'tight)

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

#create plot
plt.plot(x, y, color='blue', marker='o')
plt.title('Single Line Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.savefig('line_plot.png', dpi=300, bbox_inches = 'tight')
plt.show()
```
